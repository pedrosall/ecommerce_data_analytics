```python
import pandas as pd
import matplotlib.pyplot as plt

path = "../data/"

customers = pd.read_csv(path + "olist_customers_dataset.csv")
geolocation = pd.read_csv(path + "olist_geolocation_dataset.csv")
order_items = pd.read_csv(path + "olist_order_items_dataset.csv")
order_payments = pd.read_csv(path + "olist_order_payments_dataset.csv")
order_reviews = pd.read_csv(path + "olist_order_reviews_dataset.csv")
orders = pd.read_csv(path + "olist_orders_dataset.csv")
products = pd.read_csv(path + "olist_products_dataset.csv")
sellers = pd.read_csv(path + "olist_sellers_dataset.csv")
category_translation = pd.read_csv(path + "product_category_name_translation.csv")

tables = {
    "customers": customers,
    "geolocation": geolocation,
    "order_items": order_items,
    "order_payments": order_payments,
    "order_reviews": order_reviews,
    "orders": orders,
    "products": products,
    "sellers": sellers,
    "category_translation": category_translation,
}
```

### --- Data Quality Notes (EDA) ---
### 1. Todas las columnas de fecha en 'orders' y 'order_reviews' están cargadas como object.
###    Se convertirán a datetime en la capa de limpieza SQL (sql/02_cleaning.sql).
### 2. product_name_lenght, product_description_lenght y product_photos_qty
###    aparecen como float64 en vez de int64 debido a valores nulos presentes en 'products'.
###    Se gestionará el nulo y se tipará como entero en la capa de limpieza SQL.
### 3. Outliers identificados (los maximos son muy elevados)
###    valorar datos con mediana y 50% de los datos (std por encima de la media, y media muy alejada de la mediana)
### 4. geolocation: 261831 filas duplicadas (tiene sentido, no hay valores únicos, es situacional)
### 5. 8 pedidos tienen order_status = 'delivered' pero order_delivered_customer_date es nulo.
###    Inconsistencia real de datos (0.008% de los pedidos delivered). 
###    Se decidirá tratamiento en la capa de limpieza SQL (excluir de métricas de tiempo de entrega
###    o mantener sin fecha, a definir en sql/02_cleaning.sql).
### 6. review_comment_title (87.656 nulos, 88%) y review_comment_message (58.247 nulos, 59%)
###    tienen alta proporción de nulos, pero es esperado: la mayoría de usuarios puntúa
###    sin dejar comentario de texto. No se considera error de datos.

```python
import matplotlib.pyplot as plt

# --- Gráfico 1: Evolución mensual de revenue ---
meses = ['2016-09', '2016-10', '2016-12', '2017-01', '2017-02', '2017-03', '2017-04', 
         '2017-05', '2017-06', '2017-07', '2017-08', '2017-09', '2017-10', '2017-11', 
         '2017-12', '2018-01', '2018-02', '2018-03', '2018-04', '2018-05', '2018-06', 
         '2018-07', '2018-08']
revenue = [134.97, 40325.11, 10.90, 111798.36, 234223.40, 359198.85, 340669.68, 
           489338.25, 421923.37, 481604.52, 554699.70, 607399.67, 648247.65, 987765.37, 
           726033.19, 924645.00, 826437.13, 953356.25, 973534.09, 977544.69, 856077.86, 
           867953.46, 838576.64]

plt.figure(figsize=(12, 5))
plt.plot(meses, revenue, marker='o')
plt.xticks(rotation=45, ha='right')
plt.title('Evolución mensual del revenue (2016-2018)')
plt.xlabel('Mes')
plt.ylabel('Revenue (R$)')
plt.tight_layout()
plt.savefig('../notebooks/revenue_evolution.png')
plt.show()
```

```python
# --- Gráfico 2: Tiempo de entrega vs review score ---
rangos = ['< 5 dias', '5-10 dias', '10-15 dias', '> 15 dias']
scores = [4.37, 4.26, 4.16, 3.52]

plt.figure(figsize=(8, 5))
plt.bar(rangos, scores, color=['#2ecc71', '#f1c40f', '#e67e22', '#e74c3c'])
plt.title('Review score medio según tiempo de entrega')
plt.xlabel('Rango de entrega')
plt.ylabel('Review score medio')
plt.ylim(0, 5)
plt.tight_layout()
plt.savefig('../notebooks/delivery_vs_review.png')
plt.show()
```
